```python
# Box and Whisker Plots
columns_to_plot = ['liveness', 'duration_ms', 'danceability', 'energy',
                    'tempo', 'valence', 'loudness', 'speechiness', 'acousticness',
                    'instrumentalness']

fig, axes = plt.subplots(3, 4, figsize=(15, 15))  # Explicit 3x4 grid

sns.set_style("whitegrid")

for i, column in enumerate(columns_to_plot):
    row, col = divmod(i, 4)
    
    # Customize boxplot appearance
    df.boxplot(column=column, ax=axes[row, col], showfliers=False,
               medianprops=dict(color="#e74c3c")) # Median color
    
    axes[row, col].set_title(column, fontsize=14)

# Remove extra subplots
for i in range(len(columns_to_plot), 12):
    fig.delaxes(axes.flatten()[i])

plt.tight_layout(pad=2.0)
plt.savefig('boxplot.png')
```
